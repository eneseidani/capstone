```python
import pandas as pd

from transaction_fee_helper import (
    calculate_transaction_fees,
    transaction_cost_rate,
)


# ---------------------------------------------------------------------
# TOY PORTFOLIO
# ---------------------------------------------------------------------

portfolio_value = 1_000_000

current_weights = {
    "SPY": 0.40,
    "TLT": 0.40,
    "DGS3MO": 0.20,
}

target_weights = {
    "SPY": 0.50,
    "TLT": 0.30,
    "DGS3MO": 0.20,
}


# ---------------------------------------------------------------------
# TOY MARKET CONDITIONS
# ---------------------------------------------------------------------

fee_context = pd.DataFrame(
    {
        "Asset Class": [
            "Equity",
            "Fixed Income",
            "Cash Proxy",
        ],
        "Price": [
            500.0,
            100.0,
            float("nan"),
        ],
        "Spread": [
            0.0010,   # SPY full spread = 0.10% = 10 bps
            0.0020,   # TLT full spread = 0.20% = 20 bps
            float("nan"),
        ],
        "ADV_Dollars": [
            50_000_000,   # SPY ADV = $50m
            10_000_000,   # TLT ADV = $10m
            float("nan"),
        ],
    },
    index=[
        "SPY",
        "TLT",
        "DGS3MO",
    ],
)


# ---------------------------------------------------------------------
# CALCULATE COST
# ---------------------------------------------------------------------

total_cost, breakdown = calculate_transaction_fees(
    current_weights=current_weights,
    target_weights=target_weights,
    portfolio_value=portfolio_value,
    fee_context=fee_context,
    include_commission=True,
    return_breakdown=True,
)


cost_rate = transaction_cost_rate(
    current_weights=current_weights,
    target_weights=target_weights,
    portfolio_value=portfolio_value,
    fee_context=fee_context,
    include_commission=True,
)


print("Total transaction cost:")
print(total_cost)

print("\nTransaction cost as fraction of portfolio:")
print(cost_rate)

print("\nTransaction cost in basis points:")
print(cost_rate * 10_000)

print("\nBreakdown:")
display(breakdown)
```
